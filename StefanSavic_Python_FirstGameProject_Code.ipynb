```python
"""
Guess the Number Game

An interactive guessing game where the player selects a difficulty level,
and attempts to guess a randomly generated number within the given range.
The player receives hints after each guess and earns points based on difficulty.

Game features:
- Three difficulty levels (easy, medium, hard)
- Attempt limits based on difficulty
- Point system with difficulty multipliers
- Statistics tracking (wins, losses, score)
- Input validation for all user entries
"""

from random import randint
from IPython.display import clear_output

# defining dectionary for difficulties and their associated values
difficulties = {
    'easy': {'min': 0, 'max': 10, 'attempts': 10, 'points': 1},
    'medium': {'min': 0, 'max': 20, 'attempts': 6, 'points': 3},
    'hard': {'min': 0, 'max': 30, 'attempts': 4, 'points': 5}
}

# Global counters (outside all functions)
wins = 0
losses = 0
total_games = 0
total_score = 0


#-----------------------------------------------------------------------------------------------

def random_num_generator(min_range,max_range):
    """
    Generates a random integer between min_range and max_range (inclusive).
    
    Args:
        min_range: The minimum value for the random number
        max_range: The maximum value for the random number
    
    Returns:
        A random integer between min_range and max_range
    """
    return randint(min_range,max_range)


#-------------------------------------------------------------------------------------------------
def user_input_guess():
    """
    Prompts the user to enter a number and validates that it is a valid integer.
    Keeps asking until a valid number is entered.
    
    Returns:
        An integer representing the user's guess
    """
    while True:
        num = input('Please enter a number: ')
        try:
            num = int(num)
            return num
        except ValueError:
            print('Not a valid number, please try again')

            
#-------------------------------------------------------------------------------------------------

def select_difficulty():
    """
    Displays available difficulty levels and asks the user to select one.
    Validates the selection and returns the difficulty settings.
    
    Returns:
        A dictionary containing 'min', 'max', and 'attempts' for the selected difficulty
    """
    print('Available difficulties:', list(difficulties.keys()))
    while True:
        difficulty_choice = input('Please choose a difficulty: ')
        if difficulty_choice in difficulties:
            return difficulty_choice, difficulties[difficulty_choice] # returns the choice and dictionary associated with the selected choice

            
#---------------------------------------------------------------------------------------------------

def play_game():
    """
    Runs one complete round of the Guess the Number game.
    
    This function handles the entire game flow for a single round:
    - Prompts the user to select a difficulty level
    - Generates a random target number based on the selected difficulty
    - Allows the user to make guesses until they either:
      * Guess the correct number (win)
      * Exceed the maximum number of attempts (loss)
    - Provides feedback on each guess ("too high" or "too low")
    - Shows the number of attempts remaining
    - Updates global statistics (wins, losses, total_games, total_score)
    - Adds points to the score based on the difficulty level selected
    
    Global variables modified:
        wins: Incremented by 1 if the player guesses correctly
        losses: Incremented by 1 if the player runs out of attempts
        total_games: Incremented by 1 when a game ends (win or loss)
        total_score: Incremented by the difficulty points if player wins
    
    Returns:
        None
    """
    global wins, losses, total_games, total_score
    difficulty_choice, settings = select_difficulty()

    # Get min rangem, max range, max attempts and points from settings
    min_range = settings['min']
    max_range = settings['max']
    max_attempts = settings['attempts']
    points = settings['points']  
    
    attempts = 0
    
    target_num = random_num_generator(min_range,max_range)
    print(f"I'm thinking of a number between {min_range} and {max_range} and you have {max_attempts} guesses")
    print(f'dev test: randomly generated target number: {target_num}') #temp output for use when testing the game
    
    while True:
        if attempts < max_attempts:
            guess = user_input_guess()
            attempts += 1
            if guess == target_num:
                print('Correct')
                wins += 1
                total_games += 1
                total_score += points  # ← Use it here
                print(f'You earned {points} points!')
                break
            elif guess > target_num:
                print(f'Lower - try again, Attempts remaining: {max_attempts - attempts}')
            else:
                print(f'Higher - try again, Attempts remaining: {max_attempts - attempts}')
        else:
            print('Out of attempts')
            losses += 1
            total_games += 1
            break

            
#-------------------------------------------------------------------------------------------------------------------

#running the game
"""
Main game loop - Controls the overall flow of the Guess the Number game.

This section:
- Displays a welcome message with game instructions and hints
- Runs the main game loop that continues until the user chooses to quit
- After each game round, prompts the user to play again
- Validates user input (yes/no) and re-prompts if invalid
- Clears previous game output when the user chooses to play again
- Displays final statistics when the user chooses to quit
- Updates global statistics (wins, losses, total_games, total_score)
  for tracking player performance across multiple rounds

Flow:
1. Display welcome message and instructions
2. LOOP:
   - Run one game round
   - Ask user if they want to play again
   - Validate input (must be 'yes' or 'no')
   - If 'yes': Clear output and loop back to step 2
   - If 'no': Display final statistics and end game

Global variables tracked:
    wins: Total number of games won
    losses: Total number of games lost
    total_games: Total rounds played
    total_score: Cumulative points earned
"""
print("\nWelcome to Guess the Number!")
print("Select a difficulty, guess the secret number, and try to get it right before you run out of attempts!")
print("Hints: 'lower' = too high, 'higher' = too low\n")

while True:
    play_game()  # Shows game result
    
    # Validation loop - keeps asking until valid input
    while True:
        play_again = input('\nDo you want to play again? (yes/no): ')
        
        if play_again.lower() in ['yes', 'no']:
            break  # Valid input, exit validation loop
        else:
            print('Please enter yes or no')
    
    # Check the valid answer
    if play_again.lower() == 'yes':
        clear_output()  # Clear only when starting new game
    else:
        print('Thanks for playing!')
        break  # Exit main loop

print(f'\nWins: {wins}, Losses: {losses}, Total: {total_games}, Score: {total_score}')
```
